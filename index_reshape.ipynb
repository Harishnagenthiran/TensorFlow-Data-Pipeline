```python
import tensorflow as tf
table = tf.constant([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
print(table.numpy())
print(table.shape)
```

```python
print("first row:", table[0].numpy())
print("second row:", table[1].numpy())
print("last row:", table[-1].numpy())
print("first element:", table[0, 0].numpy())
print("last element:", table[-1, -1].numpy())
```

```python
print(table[0:2, 1].numpy())
print (table[1:1,0].numpy())
print(table.shape)
print(tf.reshape(table,[9,1]).numpy())
```

```python
a = tf.constant([[1,2],[3,4]])
b = tf.constant([[5,6],[7,8]])
print(tf.concat([a,b], axis = 1).numpy())
```

```python
stacked = tf.stack([a,b])
print(stacked.shape)
print(stacked.numpy())
```
