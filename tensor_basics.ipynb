```python
import tensorflow as tf
```

```python
scalar = tf.constant(5.0)
print (scalar)
```

```python
vector = tf.constant([1, 2, 3])
print (vector)
```

```python
matrix = tf.constant([[1, 2], [3, 4]])
print (matrix)
```

```python
tensor = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print (tensor)
```

```python
print(tensor.shape,tf.rank(tensor).numpy())
print(vector.shape,tf.rank(vector).numpy())
print(matrix.shape,tf.rank(matrix).numpy())
```

```python
a = tf.constant([1,2,3])
b = tf.constant([4,5,6])
print((a+b).numpy())
print(a-b)
print(a*b)
print(a/b)
```

```python

m1 = tf.constant([[1, 2], [3, 4]])
m2 = tf.constant([[1, 0], [0, 1]])
print((m1*m2).numpy())
```

```python
left  = tf.ones([2, 3])
right = tf.ones([3, 4])
result = tf.matmul(left, right)
print(result.numpy(), result.shape)
```
